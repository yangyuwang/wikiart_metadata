## Generate Description

```python
artist_desc_templates = {
    "gender":               "as {value}",
    "birth":                "born{year}{location}",
    "death":                "died{year}{location}",
    "education":            "educated{year}{school}",
    "nationality":          "whose nationality is {value}",
    "ethnicity":            "whose ethnicity is {value}",
    "languages_spoken":     "who spoke {value}",
    "affiliations":         "affiliated to {group}{start_year}{end_year}",
    "religion":             "believing in {religion}{start_year}{end_year}",
    "residences":           "living{location}{start_year}{end_year}",
    "visits":               "touring{location}{start_year}{end_year}",
    "interactions":         "interacting with {name}{year}{location}"
}
```

```python
import copy

with open('artist_data/demographic_information.json', 'r') as file:
    demographic_data = json.load(file)
    
data = copy.deepcopy(demographic_data["3d"])
```

```python
prepmapping = {
    "year": "in",
    "school": "in",
    "start_year": "from",
    "end_year": "to"
}
```

```python
def clean_up(dict, key, prepmapping=prepmapping):
    if key in dict.keys():
        if dict[key]:
            dict[key] = f" {prepmapping[key]} {dict[key]}"
        else:
            dict[key] = ""
```

```python
import random
import re
from collections.abc import Iterable

def description_generater(data, artist_desc_templates=artist_desc_templates, seed=42):
    """helper function to generate descriptions for paintings"""
    desc_dict = {}
    random.seed(seed)
    for k, s in artist_desc_templates.items():
        dict_format = {}

        ## Deep copy if dict or list
        if isinstance(data[k], Iterable) and not isinstance(data[k], str):
            dict_format = copy.deepcopy(data[k])
        else:
            dict_format = data[k]

        ## If none for list variables
        if k in ["education", "affiliations", "religion", "residences", "visits", "interactions"] and not dict_format:
            dict_format = {}
            for i in re.findall(r"\{([^{}]+)\}", s):
                dict_format[i] = None
        
        ## Adjust to dict for other format
        if isinstance(dict_format, list) and len(dict_format) > 0 and isinstance(dict_format[0], dict):
            dict_format = random.choice(dict_format)
        elif isinstance(dict_format, list) and len(dict_format) > 0 and not isinstance(dict_format[0], dict):
            dict_format = {"value": "and ".join(dict_format)}
        elif isinstance(dict_format, str):
            dict_format = {"value": dict_format}
        elif not dict_format:
            dict_format = {"value": None}
        
        ## Adjust location and years
        if "location" in dict_format.keys():
            if dict_format["location"]:
                if dict_format["location"]["city_full"]:
                    dict_format["location"] = " at {city_full}".format(**dict_format["location"])
                else:
                    dict_format["location"] = ""
            else:
                dict_format["location"] = ""
        
        for key in ["school", "year", "start_year", "end_year"]:
            clean_up(dict_format, key)
        
        if all(value == "" for value in dict_format.values()):
            dict_format = {key: None for key in dict_format}
        
        desc = s.format(**dict_format)
        
        if "None" not in desc:
            #print(k + ": existing, " + desc)
            desc_dict[k] = desc
        #else:
            #print(k + ": missing, " + desc)
            
    return desc_dict
```

```python
artist_desc = {}
for artist, data in demographic_data.items():
    artist_desc[artist] = description_generater(data)
```
